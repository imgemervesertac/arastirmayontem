****MP2: Literatür Taraması****

**2.1. Literatür Özeti**
Bu çalışmanın teorik altyapısı, günlük duygu durumu üzerindeki psikolojik yansımaları inceleyen akademik çalışmaların sentezlenmesiyle oluşturulmuştur. İncelenen literatür aşağıda sunulmuştur:

**-Hachenberger, Li, Realo ve Lemola (2025) -** "The association of caffeine consumption with positive affect but not with negative affect changes across the day"


Yöntem: Deneyim Örnekleme Metodolojisi (ESM) kullanılarak toplam 28.295 anket verisinin analizi.
Temel Sonuç: Kafein tüketiminin ardından pozitif duygularda belirgin bir artış gözlemlenirken, negatif duygular üzerinde tutarlı bir etki saptanmamıştır.

Kafein ve pozitif duygu arasındaki bu ilişki, uyandıktan sonraki ilk 2.5 saat içinde en güçlü seviyesine ulaşmaktadır.

**-Makki, Alharbi, Alsharif ve Aljabri (2023) -** "Caffeine Consumption and Depression, Anxiety, and Stress Levels Among University Students in Medina: A Cross-Sectional Study"

Yöntem: Taibah Üniversitesi öğrencileri üzerinde DASS-21 ölçeği kullanılarak yapılan Kesitsel Çalışma.

Temel Sonuç: Günlük kafein tüketimi ile depresyon, anksiyete ve stres şiddeti arasında anlamlı bir ilişki bulunamamıştır.

Bu bulgu, yüksek düzeyde kafein tüketiminin doğrudan psikolojik stresle ilişkili olduğu yönündeki yaygın teorilerin aksini göstermektedir.

**-Botella ve Parra (2003) -** "Coffee increases state anxiety in males but not in females"

Yöntem: Gönüllülere 3 mg ile 300 mg arasında değişen kafein dozları verilerek uygulanan deneysel yöntem ve STAI anketi.

Temel Sonuç: Kahve tüketimi, erkeklerde doza bağlı olarak durum kaygısını artırırken, kadın katılımcılarda anlamlı bir değişiklik yaratmamıştır.

Araştırma, kafeinin anksiyete üzerindeki etkisinin cinsiyete bağlı olarak değiştiğini ve kadınların kafeine karşı daha az duyarlı olabileceğini kanıtlamıştır.

**-Shanahan ve Hughes (1986) -** "Potentiation of Performance-Induced Anxiety by Caffeine in Coffe"

Yöntem: 46 gönüllü öğrenci üzerinde stresli ve stressiz görevler eşliğinde yapılan deneysel çalışma.

Temel Sonuç: Stresli durumlarda kafeinli kahve tüketimi, bireylerin kaygı puanlarını anlamlı ölçüde yükseltmektedir.

Kafein, özellikle dışsal stres faktörleri mevcut olduğunda anksiyeteyi tetikleyici bir çarpan etkisi yapmaktadır

**MP4: Analiz ve Görselleştirme**

Bu aşamada, toplanan veriler üzerinde istatistiksel analizler yapılmış ve değişkenler arasındaki ilişkiler görselleştirilmiştir.

Analiz süreci; veri temizliği, temel istatistiklerin hesaplanması ve korelasyon/dağılım grafiklerinin oluşturulmasını kapsamaktadır.

**Yapılan Analizlerin Amacı:**

Farklı duygu durumlarındaki bireylerin kafein tüketim alışkanlıklarını karşılaştırmak.

Kafein miktarının duygu durumları üzerindeki dağılımını belirlemek.

Veri setindeki uç değerleri ve anlamlı eğilimleri tespit etmek.

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# 1) Veriyi Yükleme ve Sütun Temizliği
try:
    data = pd.read_csv('data_mp3.csv', sep=";", encoding='utf-8-sig')
except:
    data = pd.read_csv('data_mp3.csv', sep=";", encoding='cp1254')

# Sütun isimlerindeki boşlukları temizle (Hata almamak için çok önemli!)
data.columns = [c.strip() for c in data.columns]
print("Veriler başarıyla yüklendi.")

# 2) Sütun Tiplerini Belirle
numeric_cols = data.select_dtypes(include=['int64', 'float64']).columns
object_cols = data.select_dtypes(include=['object']).columns

# 3) Temel İstatistikler
print("\n--- TEMEL İSTATİSTİKLER (sayısal sütunlar) ---")

if len(numeric_cols) == 0:

    print("Veri setinde sayısal sütun bulunamadı.")

else:

    for col in numeric_cols:

        print(f"Sütun: {col}")

        print("  Mean  :", data[col].mean())

        print("  Median:", data[col].median())

        print("  Std   :", data[col].std())

        print("")
print("\n--- TEMEL İSTATİSTİKLER (metinsel sütunlar) ---")

if len(object_cols) == 0:

    print("Veri setinde metinsel sütun bulunamadı.")

else:

    for col in object_cols:

        print(f"Sütun: {col}")

        print("  Unique Değer Sayısı:", data[col].nunique())

        print("  En Çok Tekrar Eden :", data[col].mode()[0])

        print("  Null Sayısı        :", data[col].isnull().sum())

        print("")

# 4) Duygu Durumuna Göre Histogramlar
# Duygu durumu sütununu otomatik bulalım
possible_cols = ['Duygu durumları', 'Duygu durumlar', 'Duygu durumu']
category_col = next((col for col in data.columns if col in possible_cols), data.columns[1])

categories = data[category_col].dropna().unique()
plt.figure(figsize=(10, 12))

for i, cat in enumerate(categories, 1):
    subset = data[data[category_col] == cat][numeric_cols[0]].dropna()
    plt.subplot(len(categories), 1, i)
    plt.hist(subset, bins=10, alpha=0.7)
    plt.title(f"Histogram: {numeric_cols[0]} — {cat}")
    plt.xlabel("Kafein Miktarı (mg)")
    plt.ylabel("Frekans")

plt.tight_layout()
plt.show()
print("Stresli → Kafein tüketim miktarı yüksektir.")

print("Nötr → Kafein tüketim miktarı dağınıktır, tepe noktası yoktur ve orta seviyelerde daha yoğundur.")

print("Mutlu → Kafein tüketimi orta seviyelerdedir, uç değerler yoktur.")

# 5) Scatter Plot
plt.figure(figsize=(7,5))
sns.scatterplot(x=category_col, y=numeric_cols[0], data=data, hue=category_col, s=100)
plt.title("Duygu Durumlarına Göre Kafein Miktarı (Scatter Plot)")
plt.show()
print("Nötr grubun diğer gruplara göre daha geniş bir miktar aralığına yayıldığı görülüyor.")

# 6) Heatmap
plt.figure(figsize=(7,5))
sns.heatmap(data[numeric_cols].corr(), annot=True, cmap="coolwarm")
plt.title("Korelasyon Heatmap")
plt.show()

print("Veri setinde sadece tek bir sayısal sütun olduğundan heatmap hiçbir ilişki göstermez, sadece kafein miktarının kendisiyle olan korelasyonu yani 1'den oluşan tek bir kare şeklinde görünür.")

# 7) Boxplot
plt.figure(figsize=(8,6))
data.boxplot(column=numeric_cols[0], by=category_col)
plt.title("Duygu Durumuna Göre Kafein Miktarı Boxplot")
plt.suptitle("") # Otomatik başlığı kaldır
plt.show()
print("Mutlu Grupta\nMedyan: Yaklaşık 85–90 mg\nDağılım: 0 mg’den yaklaşık 145 mg’ye kadar geniş bir aralık, orta seviyelerde daha yoğun.\nYorum: Mutlu kişilerde kafein tüketimi daha düşük ve daha geniş bir dağılıma sahip. Bazıları hiç tüketmezken bazıları orta düzeyde tüketiyor.\n")

print("Nötr Grupta\nMedyan: Yaklaşık 110–115 mg\nDağılım: Yaklaşık 40 mg’den 190 mg’ye kadar.\nYorum: Nötr durumda insanlar genel olarak daha fazla kafein tüketimi görülüyor. Bu durum “rutinde içilen kahve/çay” alışkanlığını yansıtıyor olabilir.\n")

print("Stresli Grup\nMedyan: Yaklaşık 95 mg\nDağılım: Yaklaşık 50 mg – 190 mg\nYorum: Stresli kişilerde kafein tüketimi artıyor, ancak bazı kişiler çok içerken bazıları sadece biraz artırıyor.\n")
```
