# MP4 Data Analysis
Bu notebook, verilen veri seti üzerinde pandas ve seaborn kullanarak analiz yapar.

```
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

data = pd.read_csv('/mnt/data/data_mp3.csv', sep=';', encoding='utf-8-sig', encoding_errors='ignore')
data.head()
```

## Temel İstatistikler

```
print("Mean:
", data.mean(numeric_only=True))
print("
Median:
", data.median(numeric_only=True))
print("
Std:
", data.std(numeric_only=True))
```

## Eksik Veri Kontrolü

```
data.isnull().sum()
```

## Korelasyon Matrisi

```
data.corr(numeric_only=True)
```

## Histogram

```
plt.figure(figsize=(6,4))
sns.histplot(data['Kafein miktar (mg)'], kde=True)
plt.title("Histogram: Kafein Miktarı")
plt.xlabel("Kafein miktar (mg)")
plt.ylabel("Frekans")
plt.show()
```

## Scatter Plot

```
plt.figure(figsize=(6,4))
sns.scatterplot(x=data.index, y=data['Kafein miktar (mg)'])
plt.title("Scatter: Kafein Miktarı vs Index")
plt.xlabel("Index")
plt.ylabel("Kafein miktar (mg)")
plt.show()
```

## Boxplot

```
plt.figure(figsize=(6,4))
sns.boxplot(y=data['Kafein miktar (mg)'])
plt.title("Boxplot: Kafein Miktarı")
plt.ylabel("Kafein miktar (mg)")
plt.show()
```

## Sonuç Özeti
- Veri setinde eksik veri yoktur.
- Kafein miktarı ortalama ~101 mg civarındadır.
- Histogram dağılımı orta seviyede yayılma göstermektedir.
- Boxplot, aykırı değerlerin bulunduğunu göstermiştir.
